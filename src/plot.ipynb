```python
import pandas as pd
import matplotlib.pyplot as plt


config = {
    "font.family": ["SimHei"],  
    "font.size": 12,  # 五号，10.5磅
    "axes.unicode_minus": False,
    "mathtext.fontset": "stix",  # 设置 LaTeX 字体，stix 近似于 Times 字体
}

plt.rcParams.update(config)
```

```python
df = pd.read_csv(r'D:\.datasets\oil_data\train\WELL_0000010\data.csv')
df.head()
```

```python
import numpy as np


INVALID_VALUES = [
    -999.25,
    -999,
    -9999
]

df = df.replace(
    INVALID_VALUES,
    np.nan
)

plt.figure(figsize=(10, 10))
# 画总池体积
plt.plot(df.index, df['总池体积(8112) m3'], label='总池体积')
plt.plot(df.index, df['活动池变化量(8126) m3'], label='活动池变化量(8126) m3')
# 出口流量(8111) %
plt.plot(df.index, df['出口流量(8111) %'], label='出口流量(8111) %')
# 全烃(8219) %
plt.plot(df.index, df['全烃(8219) %'], label='全烃(8219) %')
plt.legend()
plt.show()
```
